```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path("../data/raw")

inventory = []

for file in sorted(RAW_DIR.glob("*.csv")):
    try:
        df = pd.read_csv(file)

        inventory.append({
            "file_name": file.name,
            "rows": df.shape[0],
            "columns": df.shape[1],
            "duplicate_rows": df.duplicated().sum(),
            "missing_cells": df.isna().sum().sum(),
            "missing_pct": round(
                df.isna().sum().sum() / (df.shape[0] * df.shape[1]) * 100, 2
            ),
            "memory_mb": round(df.memory_usage(deep=True).sum() / 1024**2, 2)
        })

    except Exception as e:
        inventory.append({
            "file_name": file.name,
            "rows": None,
            "columns": None,
            "duplicate_rows": None,
            "missing_cells": None,
            "missing_pct": None,
            "memory_mb": None
        })
        print(f"Could not read {file.name}: {e}")

dataset_inventory = pd.DataFrame(inventory)

dataset_inventory
```
