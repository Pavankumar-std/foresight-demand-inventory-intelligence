```python
from pathlib import Path
import pandas as pd
import numpy as np
PROJECT_PATH = Path.cwd().parent
DATA_PATH = PROJECT_PATH / "data" / "raw"
print("Project path:", PROJECT_PATH)
print("Data path:", DATA_PATH)
```

```python
# ==========================================
# 2. LOAD CORE DATASETS
# ==========================================

sales_daily = pd.read_csv(
    DATA_PATH / "sales_daily.csv"
)

calendar = pd.read_csv(
    DATA_PATH / "calendar.csv"
)

inventory = pd.read_csv(
    DATA_PATH / "inventory_snapshots.csv"
)

sku_master = pd.read_csv(
    DATA_PATH / "sku_master.csv"
)
print("Sales daily:", sales_daily.shape)
print("Calendar:", calendar.shape)
print("Inventory snapshots:", inventory.shape)
print("SKU master:", sku_master.shape)
```

```python
datasets = {
    "sales_daily": sales_daily,
    "calendar": calendar,
    "inventory": inventory,
    "sku_master": sku_master}
for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Column names:", df.columns.tolist())
```

```python
print("SALES DAILY")
display(sales_daily.head())

print("\nCALENDAR")
display(calendar.head())

print("\nINVENTORY SNAPSHOTS")
display(inventory.head())

print("\nSKU MASTER")
display(sku_master.head())
```

```python
for name, df in datasets.items():

    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)
    print("\nData types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isna().sum())
```

```python
print("DUPLICATES")
print("-" * 40)
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()}")
print("\nDATE RANGES")
print("-" * 40)
print(
    "Sales:",
    sales_daily["Date"].min(),
    "to",
    sales_daily["Date"].max())
print(
    "Calendar:",
    calendar["date"].min(),
    "to",
    calendar["date"].max())
print(
    "Inventory:",
    inventory["Snapshot_Date"].min(),
    "to",
    inventory["Snapshot_Date"].max()
)

print("\nINVALID DATES")
print("-" * 40)

print("Sales:", sales_daily["Date"].isna().sum())
print("Calendar:", calendar["date"].isna().sum())
print("Inventory:", inventory["Snapshot_Date"].isna().sum())
```

```python
sales_skus = set(sales_daily["SKU"].unique())
inventory_skus = set(inventory["SKU"].unique())
master_skus = set(sku_master["sku_id"].unique())

print("Sales unique SKUs:", len(sales_skus))
print("Inventory unique SKUs:", len(inventory_skus))
print("Master unique SKUs:", len(master_skus))

print("\nExample Sales SKUs:")
print(sorted(sales_skus)[:10])

print("\nExample Inventory SKUs:")
print(sorted(inventory_skus)[:10])

print("\nExample Master SKUs:")
print(sorted(master_skus)[:10])

print("\nSales SKUs matching master directly:",
      len(sales_skus & master_skus))

print("Inventory SKUs matching master directly:",
      len(inventory_skus & master_skus))
```

```python
sample_skus = sorted(sales_skus)[:10]

print("Possible SKU mapping:")
for sku in sample_skus:
    converted = "SKU" + sku.replace("SKU", "").zfill(5)
    print(f"{sku}  ->  {converted}  ->  {converted in master_skus}")
```

```python
print("SALES DAILY")
print("-" * 50)

print("Negative Units Sold:",
      (sales_daily["Units_Sold"] < 0).sum())

print("Zero Units Sold:",
      (sales_daily["Units_Sold"] == 0).sum())

print("Negative Revenue:",
      (sales_daily["Revenue"] < 0).sum())

print("Negative Price:",
      (sales_daily["Price"] < 0).sum())

print("\nINVENTORY")
print("-" * 50)

print("Negative Current Stock:",
      (inventory["Current_Stock"] < 0).sum())

print("Negative On Order:",
      (inventory["On_Order"] < 0).sum())

print("Negative Lead Time:",
      (inventory["Lead_Time_Days"] < 0).sum())

print("Negative Safety Stock:",
      (inventory["Safety_Stock"] < 0).sum())

print("Negative Reorder Point:",
      (inventory["Reorder_Point"] < 0).sum())

print("Negative Inventory Value:",
      (inventory["Inventory_Value"] < 0).sum())

print("\nSKU MASTER")
print("-" * 50)

print("Negative Unit Price:",
      (sku_master["unit_price"] < 0).sum())

print("Negative Cost Price:",
      (sku_master["cost_price"] < 0).sum())
```

```python
calculated_revenue = (
    sales_daily["Units_Sold"] *
    sales_daily["Price"]
)

revenue_difference = (
    sales_daily["Revenue"] - calculated_revenue
).abs()

print("Maximum revenue difference:",
      revenue_difference.max())

print("Rows with difference > 0.01:",
      (revenue_difference > 0.01).sum())
```

```python
# Standardize SKU identifiers without modifying the raw values

sales_daily["SKU_Standard"] = (
    sales_daily["SKU"]
    .str.replace("SKU", "", regex=False)
    .astype(int)
    .astype(str)
    .str.zfill(5)
    .radd("SKU")
)

inventory["SKU_Standard"] = (
    inventory["SKU"]
    .str.replace("SKU", "", regex=False)
    .astype(int)
    .astype(str)
    .str.zfill(5)
    .radd("SKU")
)

sku_master["SKU_Standard"] = sku_master["sku_id"]

print("Sales SKU examples:")
display(sales_daily[["SKU", "SKU_Standard"]].drop_duplicates().head(10))

print("\nInventory SKU examples:")
display(inventory[["SKU", "SKU_Standard"]].drop_duplicates().head(10))

print("\nMaster SKU examples:")
display(sku_master[["sku_id", "SKU_Standard"]].drop_duplicates().head(10))
```

```python
# Validate standardized SKU mappings

sales_mapping_check = sales_daily["SKU_Standard"].isin(
    sku_master["SKU_Standard"]
)

inventory_mapping_check = inventory["SKU_Standard"].isin(
    sku_master["SKU_Standard"]
)

print("Sales SKUs not found in SKU master:",
      (~sales_mapping_check).sum())

print("Inventory SKUs not found in SKU master:",
      (~inventory_mapping_check).sum())

print("\nSales unique standardized SKUs:",
      sales_daily["SKU_Standard"].nunique())

print("Inventory unique standardized SKUs:",
      inventory["SKU_Standard"].nunique())

print("SKU master unique standardized SKUs:",
      sku_master["SKU_Standard"].nunique())
```

```python
quality_summary = pd.DataFrame({
    "Dataset": [
        "Sales Daily",
        "Calendar",
        "Inventory Snapshots",
        "SKU Master"
    ],
    "Rows": [
        len(sales_daily),
        len(calendar),
        len(inventory),
        len(sku_master)
    ],
    "Columns": [
        sales_daily.shape[1],
        calendar.shape[1],
        inventory.shape[1],
        sku_master.shape[1]
    ],
    "Duplicate Rows": [
        sales_daily.duplicated().sum(),
        calendar.duplicated().sum(),
        inventory.duplicated().sum(),
        sku_master.duplicated().sum()
    ]
})

display(quality_summary)
```

```python
# Create processed-data directory
PROCESSED_PATH = PROJECT_PATH / "data" / "processed"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)

# Save processed datasets
sales_daily.to_csv(
    PROCESSED_PATH / "sales_daily_clean.csv",
    index=False
)

calendar.to_csv(
    PROCESSED_PATH / "calendar_clean.csv",
    index=False
)

inventory.to_csv(
    PROCESSED_PATH / "inventory_snapshots_clean.csv",
    index=False
)

sku_master.to_csv(
    PROCESSED_PATH / "sku_master_clean.csv",
    index=False
)

print("Processed datasets saved successfully.")
print("\nLocation:", PROCESSED_PATH)

for file in sorted(PROCESSED_PATH.glob("*.csv")):
    print(f"{file.name} — {file.stat().st_size / 1024:.2f} KB")
```
